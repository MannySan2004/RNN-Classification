## 1. Import Libraries

```python
import tensorflow as tf  # Core TensorFlow library
# Load IMDB movie review dataset (sentiment analysis)
from keras.datasets import imdb
# Model and layers
from keras.models import Sequential  # Sequential neural network
from keras.layers import Dense       # Fully connected output layer
from keras.layers import LSTM, GRU, SimpleRNN  # Recurrent layers
from keras.layers import Embedding   # Word embedding layer
# Sequence preprocessing
from keras.utils import pad_sequences  # Padding/truncating sequences
```

## 2. Load and Preprocess Data

```python
# Load IMDB dataset
# Keep only the top 5000 most frequent words; rare words are removed
top_words = 5000
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=top_words)
# Set maximum review length (number of words per review)
max_review_length = 500
# Pad or truncate reviews to the same length
# Short reviews → padded with zeros
# Long reviews → truncated
X_train = pad_sequences(X_train, maxlen=max_review_length)
X_test = pad_sequences(X_test, maxlen=max_review_length)
```

## 3. Define RNN Model

```python
# Size of word embedding vectors
embedding_vector_length = 32
# Initialize the model
model = Sequential()
# Embedding layer Converts word indices into dense word vectors
model.add(Embedding(top_words, embedding_vector_length, input_length=max_review_length))
# LSTM layer learns long-term dependencies in text sequences
model.add(SimpleRNN(100))
#model.add(LSTM(100))
#model.add(GRU(100))
# Sigmoid activation for binary sentiment classification
model.add(Dense(1, activation='sigmoid'))
# Compile the model
# Binary crossentropy for binary classification
# Adam optimizer for efficient training
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
# Display model architecture
print(model.summary())
```

## 4. Train  Model

```python
# Train the model
model.fit(X_train, y_train,
          validation_data=(X_test, y_test),
          epochs=3,
          batch_size=64)
```

## 5. Evaluate Model

```python
# Evaluate the trained model on test set
scores = model.evaluate(X_test, y_test, verbose=0)
# Print test accuracy
print("RNN Accuracy: %.2f%%" % (scores[1]*100))
```
